```python
n=int(input("Masukan banyak mahasiswa: "))
i=0
total=0
for i in range (n):
    nilai=int(input(f"Masukan nilai ke-{i+1} mahasiswa: "))
    total+=nilai
    i+=1
print(total/n)
    
```

```python
nilai=[55,70,80,45,60,90,58]
l=0
tl=0
for n in nilai:
    if (n>=60):
        print(f"nilai {n} lulus")
        l+=1
    elif (n<60):
        print(f"nilai {n} tidak lulus")
        tl+=1
print(f"Yang lulus: {l}")
print(f"Yang tidak lulus: {tl}")
    
```
