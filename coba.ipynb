```python
i=0
data=[]
while True:
    nilai=int(input(f"Masukan nilai ke-{i+1}"))
    if (nilai<0):
        print("Nilai negatif, program dihentikan")
        break
    if (nilai>100):
        print("Nilai melebihi nilai maksimal, coba lagi")
        continue
    data.append(nilai)
    i+=1
rataRata=sum(data)/len(data)
print(rataRata)

```

```python
def subFOR(hitung):
    if hitung in range (5,10,2):
        jumlah=hitung+100
        return jumlah
    else:
        print("Nilai yang dimasukkan bukan nilai diantara 5-10 dengan selisih 2")
        return 0
total=0
while True:
        a=int(input("Masukkan angka 5-10: "))
        hasil=subFOR(a)
        if hasil==0:
             print("Looping selesai")
             break
        total+=hasil
        print(f"Nilai jumlah akhir: {total}")
```
