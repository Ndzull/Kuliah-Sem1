```python
class Distribusi:
    def __init__(self,z):
        self.z=z
    def std_norm(self):
        pi=3.15
        e=2.71
        hasil=(1/(2*pi)**0.5)*(e**-(self.z**2/2))
        return hasil
distribusi1=Distribusi(1)
distribusi2=Distribusi(2)

print(f"Hasil Distribusi dari z={distribusi1.z} adalah {distribusi1.std_norm()}")
print(f"Hasil Distribusi dari z={distribusi2.z} adalah {distribusi2.std_norm()}")
```

```python
A=[[1.1,1.2,1.3],[2.0,-2.0,2.0],[3.3,3.2,3.1]]
v=[[0.1],[0.2],[0.3]]

At=[[None,None,None],[None,None,None],[None,None,None]]

for i in range(len(A)):
    for j in range(len(A[0])):
        At[j][i]=A[i][j]

print("Matriks A Aseli:")
for baris in A:
    print(baris)

print("Matriks A Transpose:")
for baris in At:
    print(baris)

print("Matriks At dikali v:")
hasil=[[0],[0],[0]]
for i in range(len(At)):
    for j in range(len(v)):
        hasil[i][0]+=At[i][j]*v[j][0]
for baris in hasil:
    print(baris)  
```
