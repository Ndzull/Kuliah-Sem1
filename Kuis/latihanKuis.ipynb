```python
#spoil no 3
import math
a=164.70/1.3
b=154.54/1.3
c=176.64/1.3
d=143.23/1.3
e=149.38/1.3
n=5
mean=(a+b+c+d+e)/n
print(f"{mean:.2f}")
var=(((a-mean)**2)+((b-mean)**2)+((c-mean)**2)+((d-mean)**2)+((e-mean)**2))/n-1
print(f"{var:.2f}")
std=math.sqrt(var)
print(f"{std:.2f}")
```

```python
#spoil no 3
import statistics
a=164.70/1.3
b=154.54/1.3
c=176.64/1.3
d=143.23/1.3
e=149.38/1.3
angka=[a,b,c,d,e]
mean=statistics.mean(angka)
print(f"{mean:.2f}")
var=statistics.variance(angka)
print(f"{var:.2f}")
std=statistics.stdev(angka)
print(f"{std:.2f}")

```
