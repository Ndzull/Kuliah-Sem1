```python
for angka in range(2,31):
    prima=True
    for pembagi in range(2,angka):
        if angka%pembagi==0:
            prima=False
            break
    if prima:
        print(angka)
```

```python
berat=[56,58,61,59,120,60,62]
n=1
for i in berat:
    if(i>100):
        print(f"Outlier ditemukan pada data ke-{n}: {i}")
        break
    else:
        print(f"Data ke-{n}: {i}")
    n+=1
```
