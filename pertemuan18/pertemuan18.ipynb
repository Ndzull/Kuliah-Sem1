```python
class Kucing:
    def __init__(self, nama, warna):
        self.nama=nama
        self.warna=warna 
    def miaw(self):
        print(f"{self.nama} mengatakan: miaw")
kucing1=Kucing("Milo","Oranye")
kucing2=Kucing("Yanto","Hitam")

print(f"Usia kucing 1 bernama {kucing1.nama}, berwarna {kucing1.warna}")
print(f"Usia kucing 2 bernama {kucing2.nama}, berwarna {kucing2.warna}")

kucing1.miaw()
kucing2.miaw()
```

```python
class Identitas:
    def __init__(self, nama, nrp):
        self.nama=nama
        self.nrp=nrp 
    def halo(self):
        print(f"{self.nama} mengatakan: halo, NRP saya {self.nrp}")
identitas1=Identitas("Ijul",2043251046)
identitas2=Identitas("Hani",2043251056)

identitas1.halo()
identitas2.halo()
```

```python
class InfoKos:
    def __init__(self, nama, wilayah):
        self.nama=nama
        self.wilayah=wilayah
    def info(self):
        print(f"Halo, aku {self.nama}, aku ngekos di daerah {self.wilayah}")

infoKos1=InfoKos("Ijul", "Mulyos")
infoKos2=InfoKos("Grace", "Gebang")
infoKos3=InfoKos("Hani", "Keputih")

infoKos1.info()
infoKos2.info()
infoKos3.info()
```

```python
class RobotJalan:
    def berjalan(self):
        print("Robot sedang berjalan di darat")
        
class RobotTerbang:
    def terbang(self):
        print("Robot sedang terbang di udara")
        
class RobotSerbaGuna:
    def __init__(self):
        self.perilakuJalan = RobotJalan()
        self.perilakuTerbang = RobotTerbang()
        
    def perilakuGabungan(self):
        print("Memulai perilaku serbaguna:")
        self.perilakuJalan.berjalan()
        self.perilakuTerbang.terbang()
        print("Perilaku serbaguna selesai")

mesinJalan = RobotSerbaGuna()
mesinJalan.perilakuGabungan()
```
