```python
#cara panjang cik
def add(a,b):
    result=[]
    for first, second in zip(a,b):
        result.append(first+second)
    return result

add([1,2,3],[4,5,6])
```

```python
import numpy as np

nilaiSiswa=np.array([55,43,98,76,65,77,65,90])

print(f"Nilai median: {np.median(nilaiSiswa)}")
print(f"Nilai minimal: {np.min(nilaiSiswa)}")
print(f"Nilai maksimal: {np.max(nilaiSiswa)}")
print(f"Nilai rata-rata: {np.average(nilaiSiswa)}")
print(f"Total nilai: {np.sum(nilaiSiswa)}")
print(f"Standar deviasi: {np.std(nilaiSiswa)}")
```

```python
import numpy as np
data=np.array([102,105,110,107,111,115,120])

print("Saham selama 7 hari")
print(data)
print(f"Rata-rata saham: {np.average(data)}")
print(f"Perubahan saham harian: {np.diff(data)}")
print(f"Volatilitas saham: {np.std(data):.2f}")

```

```python
import numpy as np

a=np.matrix([[10,12],
             [8,15],
             [20,5]])
b=np.matrix([[5,7],
             [6,2],
             [10,3]])
total=a+b
print("Matriks A tambah matriks B")
print(total)
print("Total pengiriman harian") #baris
print(np.sum(total, axis=1))
print("Total pengiriman ke setiap tujuan") #kolom
print(np.sum(total, axis=0))
```

```python
import numpy as np
a=np.array([[3,1],
            [2,4]])
b=np.array([29,38])
x=np.linalg.solve(a,b) #tadi nyoba kalau datanya matrix gx iso
print(f"Jumlah paket tipe 1: {x[0]}")
print(f"Jumlah paket tipe 2: {x[1]:.1f}")

```
