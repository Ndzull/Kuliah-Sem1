```python
class Mobil:
    def __init__(self):
        self.mesin=Mesin()
        self.lampu=Lampu(2)
    def berjalan(self):
        self.mesin.mulai()
        self.lampu.nyala()
        print("Mobil berjalan")
class Mesin:
    def mulai(self):
        print("Mesin dinyalakan")
class Lampu:
    def __init__(self, jumlah_lampu):
        self.jumlah_lampu=jumlah_lampu
    def nyala(self):
        print("Lampu dinyalakan")
        print(f"Jumlah lampu yang dinyalakan: {self.jumlah_lampu}")

mobil=Mobil()
mobil.berjalan()
```
