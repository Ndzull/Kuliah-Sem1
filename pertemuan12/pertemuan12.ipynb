```python
nama=input("Masukan nama: ")
nilai=int(input("Masukan nilai: "))
if(nilai>=85):
    print(f"Selamat {nama}, nilai anda A")
elif(nilai>=75 and nilai<85):
    print(f"Selamat {nama}, nilai anda B")
elif(nilai>=60 and nilai<75):
    print(f"Semangat {nama}, nilai anda C")
elif(nilai<60):
    print(f"Lebih giat lagi {nama}, nilai anda D")
```

```python
x=int(input("Masukan nilai X: "))
y=int(input("Masukan nilai Y: "))
print("Operasi (X+Y)*(X-Y)")
hasil=(x+y)*(x-y)
print(f"Operasi ({x}+{y})*({x}-{y})= {hasil}")
if(hasil%2==0):
    print("Hasil adalah genap")
else:
    print("Hasil adalah ganjil")
```
