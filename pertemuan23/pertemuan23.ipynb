```python
class Distribusi:
    def __init__(self):
        self.e=2.71
        self.lamda=3
    def hitung(self,x):
        self.x=x
        hasil=self.lamda*(self.e**(-self.lamda*self.x))
        return hasil
    
peluang=Distribusi()
peluang.hitung(1)
```
