```python
class Belanja:
    def __init__(self, daftar_barang, saldo):
        self.daftar_barang = daftar_barang
        self.saldo = saldo

    def hargaTotal(self):
        total = 0
        for nama, harga, jumlah in self.daftar_barang:
            total += harga * jumlah
        return total

    def ceksaldo(self):
        print("Rincian Belanja")
        for nama, harga, jumlah in self.daftar_barang:
            total_barang = harga * jumlah
            print(f"{nama}: {jumlah} x {harga} = {total_barang}")

        print("Hasil Pengecekan")
        total_semua = self.hargaTotal()

        if self.saldo >= total_semua:
            print(f"Saldo cukup! Total belanja: {total_semua}")
        else:
            print(f"Saldo kurang. Total belanja: {total_semua}, Saldo: {self.saldo}")
            print("\nBarang yang masih bisa dibeli:")

            for nama, harga, jumlah in self.daftar_barang:
                jumlah_bisa = self.saldo // harga
                if jumlah_bisa > 0:
                    print(f"- {nama} bisa dibeli sebanyak {jumlah_bisa}")
                else:
                    print(f"- {nama} tidak bisa dibeli sama sekali")

daftar = [
    ("Sate", 1000, 11),
    ("Nasi Goreng", 15000, 2),
    ("Es Teh", 3000, 5)
]

belanja = Belanja(daftar, saldo=10000)
belanja.ceksaldo()

```

```python
class Barang:
    def __init__(self, nama, harga):
        self.nama = nama
        self.harga = harga

    def __repr__(self):
        return f"{self.nama} (Rp{self.harga})"

class KeranjangBelanja:
    def __init__(self):
        self.items = []

    def tambah(self, barang):
        self.items.append(barang)
        print(f"{barang.nama} ditambahkan ke keranjang.")

    def hapus_terakhir(self):
        if self.items:
            barang = self.items.pop()
            print(f"{barang.nama} dihapus dari keranjang.")
        else:
            print("Keranjang sudah kosong.")

    def total_harga(self):
        return sum(b.harga for b in self.items)

    def tampilkan(self):
        if not self.items:
            print("Keranjang kosong.")
        else:
            print("Isi keranjang:")
            for b in self.items:
                print("-", b)

class User:
    def __init__(self, nama, saldo):
        self.nama = nama
        self.saldo = saldo
        self.keranjang = KeranjangBelanja()

    def tambah_barang(self, barang):
        self.keranjang.tambah(barang)

    def checkout(self):
        total = self.keranjang.total_harga()
        print(f"\nTotal belanja: Rp{total}")
        print(f"Saldo kamu: Rp{self.saldo}")

        if total <= self.saldo:
            self.saldo -= total
            print("Pembelian berhasil")
            print(f"Sisa saldo: Rp{self.saldo}")
            self.keranjang.items = []
        else:
            print("Saldo tidak cukup")
            print("Menghapus barang terakhir...")
            self.keranjang.hapus_terakhir()


if __name__ == "__main__":
    b1 = Barang("Indomie", 3000)
    b2 = Barang("Telur", 12000)
    b3 = Barang("Daging Ayam", 35000)

    user = User("Ndzull", 40000)
    user.tambah_barang(b1)
    user.tambah_barang(b2)
    user.tambah_barang(b3)

    print()
    user.keranjang.tampilkan()
    user.checkout()
```
