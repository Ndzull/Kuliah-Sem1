```python
target=5000000
tabunganRafi=0
pemasukan=200000
tambahan=0
n=1
while(tabunganRafi<target):
    if(n>=2):
        tambahan=50000*(n-1)
        tabunganRafi+=pemasukan+tambahan
    else:
        tabunganRafi+=pemasukan
    print(f"Jumlah tabungan minggu ke-{n}: {tabunganRafi}")
    print(f"Jumlah tambahan minggu ke-{n}: {pemasukan+tambahan}")
    print()
    n+=1
```

```python
kata=input("Masukan kata: ")
huruf=list(kata)
print(huruf)
hurufVokal=["a","i","u","e","o","A","I","U","E","O"]
countHurufVokal=0

for i in huruf:
    if i in hurufVokal:
        countHurufVokal+=1
    else:
        countHurufVokal+=0
print(f"Jumlah huruf vokal yang ada di masukan tersebut adalah: {countHurufVokal}")
```
