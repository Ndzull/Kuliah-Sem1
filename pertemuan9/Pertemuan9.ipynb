```python
l=((5,2,3),(4,7,6),(8,9,6))
lList=[]
print(l)
for sl in l:
    sl=list(sl)
    lList.append(sl)

print(lList)
n=0
for i in range(len(l)):
    del lList[n][-1]
    lList[n].append(10)
    n+=1

n=0
for i in range(len(l)):
    lList[n]=tuple(lList[n])
    n+=1

print(lList)

```
